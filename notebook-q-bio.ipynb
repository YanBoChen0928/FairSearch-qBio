```python
import json

# Store papers for each category separately
pe_papers = []
qm_papers = []
gn_papers = []

path = '/kaggle/input/datasets/organizations/Cornell-University/arxiv/arxiv-metadata-oai-snapshot.json'

with open(path, 'r') as f:
    for line in f:
        paper = json.loads(line)
        cats = paper.get('categories', '')
        
        # Each paper goes into its own category bucket
        if 'q-bio.PE' in cats:
            pe_papers.append(paper)
        if 'q-bio.QM' in cats:
            qm_papers.append(paper)
        if 'q-bio.GN' in cats:
            gn_papers.append(paper)

# Summary
print(f"q-bio.PE: {len(pe_papers)} papers")
print(f"q-bio.QM: {len(qm_papers)} papers")
print(f"q-bio.GN: {len(gn_papers)} papers")
print(f"Combined total: {len(pe_papers) + len(qm_papers) + len(gn_papers)} papers")
```
